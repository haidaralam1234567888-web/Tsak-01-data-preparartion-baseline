```python

# Task 01: Data Preparation and Baseline

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import pandas as pd

# Load the public Iris dataset
iris = load_iris(as_frame=True)
df = iris.frame

# Separate features and target
X = df.drop(columns=["target"])
y = df["target"]

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dataset information
print("Dataset shape:", df.shape)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Features:", list(X.columns))
print("Target classes:", list(iris.target_names))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

# Create a simple baseline model
baseline = DummyClassifier(strategy="most_frequent")

# Train the baseline model
baseline.fit(X_train, y_train)

# Make predictions on test data
baseline_predictions = baseline.predict(X_test)

# Calculate baseline accuracy
baseline_accuracy = accuracy_score(
    y_test, baseline_predictions
)

print("Baseline Accuracy:", round(baseline_accuracy * 100, 2), "%")
```

```python

# Step 4: Train a Machine Learning Model

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Create the model
model = LogisticRegression(max_iter=200)

# Train the model using training data
model.fit(X_train, y_train)

# Predict on test data
predictions = model.predict(X_test)

# Calculate model accuracy
model_accuracy = accuracy_score(y_test, predictions)

# Display baseline and model accuracy
print("Baseline Accuracy:", round(baseline_accuracy * 100, 2), "%")
print("Model Accuracy:", round(model_accuracy * 100, 2), "%")

# Display classification report
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        predictions,
        target_names=iris.target_names
    )
)
```


## Results and Conclusion

The Iris dataset was divided into training and testing sets using an 80:20 split.

A DummyClassifier was used to establish the baseline accuracy. Logistic Regression was then trained and evaluated on the test set.

- Baseline accuracy: 33.33%
- Logistic Regression accuracy: 96.67%

The Logistic Regression model performed better than the baseline on the test dataset. It correctly classified 29 out of 30 test samples.

The results show that a simple machine learning model can classify Iris flower species effectively on this test split.
